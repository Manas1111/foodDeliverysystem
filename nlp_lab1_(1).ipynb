```python
import nltk
nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('words')

import string
import pandas as pd
from nltk.corpus import brown, stopwords, words
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.util import ngrams
from collections import Counter
from nltk.metrics.distance import edit_distance
```

## Q1. Tokenization and Stop Word Removal

```python
text = ' '.join(brown.words()[:2000])
sentences = sent_tokenize(text)[:20]

stop_words = set(stopwords.words('english'))
punct = set(string.punctuation)

cleaned_data = []
for sent in sentences:
    word_tokens = word_tokenize(sent)
    lower_tokens = [w.lower() for w in word_tokens]
    cleaned = [w for w in lower_tokens if w not in punct and w not in stop_words and w.isalpha()]
    cleaned_data.append((sent, word_tokens, cleaned))

for i in range(5):
    print("\nOriginal Sentence:", cleaned_data[i][0])
    print("Word Tokens:", cleaned_data[i][1])
    print("Cleaned Tokens:", cleaned_data[i][2])
```

## Q2. Stemming and Lemmatization

```python
alpha_words = [w.lower() for w in brown.words() if w.isalpha()][:30]
extra_words = ['playing', 'studies', 'running', 'better', 'cars']
final_words = list(dict.fromkeys(alpha_words + extra_words))

ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stems = [ps.stem(w) for w in final_words]
lemmas = [lemmatizer.lemmatize(w) for w in final_words]

df_q2 = pd.DataFrame({'Original': final_words, 'Stemmed': stems, 'Lemmatized': lemmas})
df_q2
```

## Q3. Spelling Correction

```python
correct_words = words.words()

misspelled = ['recieve', 'adress', 'definately', 'wich', 'thier',
              'occured', 'begining', 'seperate', 'goverment', 'accomodate']

corrections = {}
for word in misspelled:
    candidates = [(edit_distance(word, w), w) for w in correct_words if w[0].lower() == word[0]]
    best = min(candidates, key=lambda x: x[0])[1]
    corrections[word] = best

print("Misspelled -> Suggested Correction")
for word, correction in corrections.items():
    print(word, "->", correction)
```

```python
test_sentences = [
    "I will recieve the parcel tomorrow",
    "Please send it to my adress",
    "I will definately come to the party",
    "I dont know wich one is correct",
    "This is thier house"
]

print("Corrected Sentences:")
for sent in test_sentences:
    words_in_sent = sent.split()
    corrected_sent = [corrections.get(w.lower(), w) for w in words_in_sent]
    print("Original :", sent)
    print("Corrected:", ' '.join(corrected_sent))
    print()
```

## Q4. N-gram Analysis

```python
brown_words = [w.lower() for w in brown.words()[:5000] if w.isalpha()]

unigrams = list(ngrams(brown_words, 1))
bigrams = list(ngrams(brown_words, 2))
trigrams = list(ngrams(brown_words, 3))

unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print("Top 10 Unigrams:")
for gram, freq in unigram_freq.most_common(10):
    print(gram, freq)

print("\nTop 10 Bigrams:")
for gram, freq in bigram_freq.most_common(10):
    print(gram, freq)

print("\nTop 10 Trigrams:")
for gram, freq in trigram_freq.most_common(10):
    print(gram, freq)
```

```python
print("Unigrams show individual word frequency with no context.")
print("Bigrams capture relationship between two consecutive words giving local context.")
print("Trigrams capture three word sequences giving richer contextual and syntactic information.")
print("As N increases, more context is captured but data becomes sparser and needs a larger corpus.")
```
